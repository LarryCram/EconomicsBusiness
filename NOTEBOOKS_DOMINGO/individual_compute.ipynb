### This notebook will use the datalist of to the individual citation reports and compute scores of researchers

```python
import pandas as pd
import numpy as np
import pickle
```

```python
journals = pd.read_excel('RESULTS/fractional_inst_journal_scores.xlsx',sheet_name='journals')
institutions = pd.read_excel('RESULTS/fractional_inst_journal_scores.xlsx',sheet_name='institutions')
resdata = pd.read_excel('DATAFILES/researchers_sample.xlsx',sheet_name='researchers')
forscores = journals[['eco_incites','journal_score']].copy()
iforscores = institutions[['eco_institution','iscore']].copy()
```

### Uploading the pkl files

```python
researchers = pd.read_pickle("HOME_PICKLE_FILES/researchers_list.pkl")
datalist = pd.read_pickle("HOME_PICKLE_FILES/resdatalist.pkl")
```

```python
acumulador = []
for k,res in zip(range(len(datalist)),researchers):
    df = datalist[k]
    data = df.merge(forscores, on='eco_incites',how='left')
    scores = data.merge(iforscores, on='eco_institution',how='left')
    scores['iscore'] = scores['iscore'].fillna(1)
    scores['journal_score'] = scores['journal_score'].fillna(.1)
    suma = (scores['journal_score']*scores['iscore']/scores['ninst']).sum()
    acumulador.append([res,suma])
acumdata =  pd.DataFrame(acumulador, columns=['ACR','suma'])
fdata = acumdata.merge(resdata, on='ACR',how='left')
fdata['coc'] = fdata['suma']/fdata['PUB']
fdata['score'] = 100*np.sqrt(fdata['coc']/ max(fdata['coc']))
columnas = ['NAME','Research_Profile','ACR', 'PUB', 'CIT','HCP','suma','coc', 'score', 'Group']
prodata = fdata[columnas].copy()
prodata.sort_values('score',ascending=False, inplace=True)
prodata.to_excel("RESULTS/researchers_results.xlsx")
```
